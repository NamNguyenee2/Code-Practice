### Hashsets

```python
# Hashsets

s = set()
print(s)
```

```python
s.add(1)
print(s)
```

```python
# Add item into Set - O(1)

s.add(2)
s.add(3)
print(s)
```

```python
print(len(s))
s.remove(1)
print(s)
```

```python
s.add('a')
print(s)
```

```python
# Look up if item in set - O(1)
if 'a' in s:
    print(True)
```

```python
string = "aaaaaaaaaaaaccccccccccccccdddddddddddddtttttttttttttt"

set_str = set(string) # Set O(N) - N is the length of the string
print(set_str)
```

```python
# Loop over term in a set - O(n)
for i in set_str:
    print(i)
```

### Hashmaps - Dictionaries

```python
d = {'Nam': 1, 'Tran': 2, 'Nguyen':3}
print(d)
```

```python
# Add key:val in dictionary: O(1)

term_add = {'EE': 2}
# d[term_add.keys()[0]] = term_add.values()[0]
print(type(term_add.values()))
# print("New dictionary:", d)

d['EE'] = 2
print("new d", d)
```

```python
# Check for presense of key in dictionary: O(1)
if 'Nam' in d:
    print(True)
```

```python
# Check the value corresponding to a key in the dictionary: O(1)

# Remember to check if the key exists first.
print(f"Value of the key {'Nam'}:", d['Nam'])
```

```python
# Look over the key:val in the dictionary - O(N)
for key, val in d.items():
    print(f"Key {key} and {val}")
```

### Defaultdic

```python
from collections import defaultdict

default = defaultdict()
```

```python
print(default)
```

```python
default = defaultdict(int)
print(default)

default[2]
print(default)
```

```python
print(default[3])
print("New default:", default)
```

```python
default = defaultdict(list)
print(default)

default[2]
print(default)
```

```python
default["a"]
```

```python
print(default)
```
