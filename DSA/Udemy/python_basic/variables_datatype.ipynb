## Variables

```python
## Dynamic typing
## Python allows the type of a variable to change as the program executes

var = 10 # int
print(var, type(var))

var = 'Hello'
print(var, type(var))

var = 3.14
print(var, type(var))
```

```python
age = int(input('What is the age:'))
print(age, type(age))
```
