## Basis Syntaxx Rules in Python

```python
### Case sensitivity
## this is single line comment
'''
This is an example of multiline comment
'''

## Sensitive cases
name = 'Nam'
Name = 'Nguyen'

print(name)
print(Name)
```

```python
## Indentation
## Python uses indentation to define blocks of code.
## Consistent use of spaces (commonly 4) or a tab is required

age = 32

if age > 30:
    print("You are older than 30.")

## This is a single line
print('Hello, World!')
```

```python
## Line continuation
## Use a backslash '\' to continue a statement to the next line.

total = 1 + 2 + 3 + 4 \
        + 5 + 6 + 7 + 8

print(total)
```

```python
## Multiple statements on a single line, use ';'
x = 5; y = 10; z = x + y

print(x, y, z)
```
