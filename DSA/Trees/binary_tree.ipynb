```python

# Binary trees

class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    
    def __str__(self):
        return str(self.val)

```

```python
#       1
#   2       3
# 4     5 10

A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)
```

```python
A.left  = B
A.right = C
B.left  = D
B.right = E
C.left  = F
```

```python
# Recursive pre order traversal (DFS) Time: O(n), Space: O(n)

def pre_order(node):
    if not node:
        return
    
    print(node)
    pre_order(node.left)
    # print(node)
    pre_order(node.right)


def in_order(node):
    if not node:
        return
    
    in_order(node.left)
    print(node)
    in_order(node.right)

def pos_order(node):
    if not node:
        return
    
    pos_order(node.left)
    pos_order(node.right)
    print(node)
```

```python
# pre_order(A)
# in_order(A)
pos_order(A)
```

```python
# Iterative pre order traversal (DFS), Time: O(n), Space: O(n)
def pre_order_iterative(node):
    stk = [node]

    while stk:
        node = stk.pop()
        print(node)
        if node.right: stk.append(node.right)
        if node.left: stk.append(node.left)
    
pre_order_iterative(A)
```

```python
# Level order traversal (BFS), Time: O(n), Space: O(n)

from collections import deque

def level_order(node):
    q = deque()
    q.append(node)

    while q:
        node = q.popleft()
        print("node", node)
        if node.left: q.append(node.left)
        if node.right: q.append(node.right)

level_order(A)
```
