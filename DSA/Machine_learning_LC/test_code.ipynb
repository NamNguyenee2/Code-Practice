```python
import numpy as np
def mse(y_true, y_pred):
    sum = []
    for i, j in zip(y_true, y_pred):
        sum.append((i -j)**2)
    print('sum', sum)
    return np.mean(sum)

y_true = [1, 2, 3]
y_pred = [1, 2, 3]

mse(y_true, y_pred)
```

```python
a, b = [[1,2],[3,4]], [[4,3],[2,1]]

print(a[0][1])
print(b[0][1])
def matrix_add(a, b):
    rows, cols = len(a), len(a[0])
    c_matrix = a
    for r in range(rows):
        for c in range(cols):
            c_matrix[r][c] = a[r][c] + b[r][c]
    return c_matrix

c = matrix_add(a, b)
print(c)
```

```python
def percentile(values, p):
    n = len(values)
    pos = divmod(p*(n-1), 100)
    pos = list(pos)

    if pos[1] == 0:
        return values[pos[0]]
    
    else:
        quo, res = pos[0], pos[1]
        value_pos = res * (values[quo+1] - values[quo])/100 + values[quo]
        print()
        return value_pos

percentile([1,2,3,4,5, 6], 50)
```

```python
def matrix_vector_multiply(matrix, vector):
    rows, cols = len(matrix), len(matrix[0]) # cols = rows_vector
    vector_result = []
    for r in range(rows):
        sum_up = 0
        for c in range(cols):
            sum_up += matrix[r][c] * vector[c]
        vector_result.append(sum_up)

    return vector_result




matrix_vector_multiply([[1,2],[3,4]], [2,1])
```

```python
import math

import math

def logistic_predict(features, weights, bias):
    n = len(features)
    output = n * [0]

    for i in range(n):
        value = [0] * len(features[0])
        for j in range(len(features[0])):
            value[j]  = features[i][j] * weights[j]
            
        output_hid = sum(value) + bias
        output[i]  = 1 / (1 + math.exp(- output_hid)) 

    return output



logistic_predict([[0,0],[1,1]], [1,1], 0)
```
