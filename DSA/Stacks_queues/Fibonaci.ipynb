```python
## Finbonacci
## Time complexity O(2^n)
## Space complexity O(n)

def F(n):
    if n == 0:
        return 0

    elif n == 1:
        return 1
    
    else:
        return F(n-2) + F(n-1)
```

```python
F(5)
```

```python
class SinglyNode:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next

    def __str__(self):
        return str(self.val)
    
Head = SinglyNode(1)
```

```python
print(Head)
A = SinglyNode(3)
B = SinglyNode(4)
C = SinglyNode(7)

Head.next = A

print("Head:", Head)
print("A:", A)
```

```python
print(Head.next)
A.next = B
B.next = C
```
