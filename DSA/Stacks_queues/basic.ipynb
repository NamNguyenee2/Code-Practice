### Queue - First in first out

```python
from collections import deque
```

```python
queue = deque()
```

```python
print(queue)
```

```python
queue.append(1)
queue.append(2)
queue.append(3)
print(queue)
```

```python
# popleft
queue.popleft()
```

```python
queue
```

```python
# peek from left and right side
peek_left = queue[0]
peek_right = queue[-1]

print("left", peek_left)
print("right", peek_right)
```
