```python
def gcdOfStrings(str1: str, str2: str) -> str:
    word1 = str1
    word2 = str2
    l_1 = len(word1)
    l_2 = len(word2)

    consider_range = list(range(1, l_2 // 2 + 1)) + [l_2]

    for len_pattern in consider_range:
        if l_2 % len_pattern == 0:
            pattern = word2[:len_pattern]
            pattern_check_1 = pattern * int(l_1/len_pattern)
            pattern_check_2 = pattern * int(l_2/len_pattern)

            if (pattern_check_1 == word1) and (pattern_check_2 == word2):
                return pattern

        else:
            return ''


gcdOfStrings('ABCABC', 'ABC')
```

```python

def minIncrease(nums):
    length = len(nums)
    num_special, exclude_index = check_special(nums)

    list_special = []
    list_action  = []

    for i in range(1, length - 1):
        if i not in exclude_index:
            new_list = nums.copy()
            num_special_new, num_action = do_action(new_list, i, num_special)
            list_special.append(num_special_new)
            list_action.append(num_action)
    
    if list_special: 
            max_special = max(list_special)
            if max_special == num_special:
                return 0
            index = [i for i, value in enumerate(list_special) if value == max_special]

            min_action = min([list_action[i] for i in index])

            return min_action
            
    return 0

def check_special(nums):
    count         = 0
    exclude_index = []
    for i in range(1, len(nums) - 1):
        if (nums[i] > nums[i - 1]) and (nums[i] > nums[i + 1]):
            count += 1
            exclude_index.append(i)
    return count, exclude_index

def do_action(new_list, index, num_special):
    num_action = 0

    while True:
        new_list[index]     += 1
        num_special_new, _  = check_special(new_list)
        num_action          += 1

        if (new_list[index]  > new_list[index - 1]) and (new_list[index]  > new_list[index + 1]):
            break

    return num_special_new, num_action
```

```python
nums = [17,3,10,4,19]


minIncrease(nums)
```

```python
def shuffle(nums, n):
        output = nums

        for i in range(n):
            output[2*i]     = nums[i]
            output[2*i + 1] = nums[i + n]
        
        return output

nums = [1,1,2,2]
n = 2

shuffle(nums, n)
```

```python
from collections import Counter

nums = [1,5,3,2,2,7,6,4,8,9]

def findErrorNums(nums):
    nums.sort()
    n = len(nums)
    ref_list = list(range(1, n + 1))
    c = Counter(nums).most_common(1)[0][0]
    
    return [c, list(set(nums) ^ set(ref_list))[0]]
```

```python
findErrorNums(nums)
```

```python
def reverse(x):
        x_str = str(x)
        x_str_list = list(x_str)
        n = len(x_str_list)
        output = []

        if x_str_list[0] == '-':
            for i in reversed(range(1, n)):
                output.append(x_str_list[i])
            
            output = ''.join(output)
            return -int(output)
        
        else:
            for i in reversed(range(0, n)):
                output.append(x_str_list[i])
       
            output = ''.join(output)

            return int(output)

x = 123
print(reverse(x))
```
