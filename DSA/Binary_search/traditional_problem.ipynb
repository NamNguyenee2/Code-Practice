```python
# Given a sorted list 
# Need to check if number is in list

def binary_search(ls, number):
    N = len(ls)
    L = 0
    R = N - 1

    while L <= R:
        M = L + (R-L)//2
        if number == ls[M]:
            return True
        elif number > ls[M]:
            L = M + 1
        else:
            R = M - 1
    
    return False
    
ls = [-3, -1, 0, 3, 5]

number = 2
```

```python
binary_search(ls, number)
```
