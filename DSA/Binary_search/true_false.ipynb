```python
def find_day(array):
    N = len(array)
    L = 0
    R = N - 1

    while L <= R:
        M = L + (R-L)//2
        if M == L:
            return M+1
        else:
            if array[M] == True:
                L = M
            else:
                R = M
        # print(1)
    return None


array_1 = [True, True, False, False]
array_2 = [True, True, True, True, False, False]
print("test case 1", find_day(array_1))
print("test case 2", find_day(array_2))
```
